```python
# Annotate = it is use to display the codinate of x and y axis
```

```python
plt.figure(figsize=(12,6), facecolor="#76ff1a")
plt.plot(x,y,label='X-axis values', color="#ff1a53",linewidth=4,marker='*',
         markersize=10,markerfacecolor='green', 
         markeredgecolor='black',linestyle='--')
plt.xlabel('X-axis',fontsize=18)
plt.ylabel('Y-axis',fontsize=18)
plt.title('Graph',fontsize=20,color='grey')
plt.gca().set_facecolor("#1aa3ff")
plt.grid(True)
plt.legend(title='Values',fontsize=12,loc='best')  # loc='upper center', 'lower center', 'upper left'
# Annotate()
for i in range(len(x)):
    plt.annotate(f'({x[i]},{y[i]})',(x[i],y[i]),va='top',ha='left',fontsize='12')  # va='top/bottom/baseline' ha='left'/'right'
```

```python
x=np.arange(1,11)
y=x**2
z=x**3
```

```python
plt.plot(x,y,z, color="#ff1a66")
```

```python
plt.figure(figsize=(12,6), facecolor="#76ff1a")
plt.plot(x,y,label='X-axis values', color="#ff1a53",linewidth=4,marker='*',
         markersize=10,markerfacecolor='green', 
         markeredgecolor='black',linestyle='--')

plt.plot(x,z,label='X-axis values', color="#ff1a53",linewidth=4,marker='*',
         markersize=10,markerfacecolor='green', 
         markeredgecolor='black',linestyle='--')

plt.xlabel('X-axis',fontsize=18)
plt.ylabel('Y-axis',fontsize=18)
plt.title('Graph',fontsize=20,color='grey')
plt.gca().set_facecolor("#1aa3ff")
plt.grid(True)
plt.legend(title='Values',fontsize=12,loc='best')  # loc='upper center', 'lower center', 'upper left'
# Annotate()
for i in range(len(x)):
    plt.annotate(f'({x[i]},{y[i]})',(x[i],y[i]),va='top',ha='left',fontsize='12')  # va='top/bottom/baseline' ha='left'/'right'
    plt.annotate(f'({x[i]},{z[i]})',(x[i],z[i]),va='bottom',ha='right',fontsize='12')
```
